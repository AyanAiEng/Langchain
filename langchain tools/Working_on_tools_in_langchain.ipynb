# ***Using tools decorator***

```python
from langchain_core.tools import tool
```

```python
# Step 1 - create a function

def multiply(a,b):
    '''
    mulitplying two numbers
     '''
    return a*b
```

```python
# Step 2 - add type hints

def multiply(a: int, b:int) -> int:
    """Multiply two numbers"""
    return a*b
```

```python
# Step 3 - add tool decorator

@tool

def multiply(a: int, b:int) -> int:
    """Multiply two numbers"""
    return a*b
```

```python
multiply_tool = multiply.invoke({'a':6,"b":9})

print(multiply_tool)
```

```python
print(multiply.name)
print(multiply.args)
print(multiply.description)
```

# ***Method 2 - Using StructuredTool***

```python
from langchain_core.tools import StructuredTool
from pydantic import BaseModel,Field
```

```python
class Multiply_input(BaseModel):
    a : int = Field(required = True,description="The frist number to add")
    b : int = Field(required = True,description="The second number to add")
```

```python
def multiply_fuu(a,b):
    return a * b
```

```python
multiply_tool = StructuredTool.from_function(
    func=multiply_fuu,
    args_schema=Multiply_input,
    description="Multiply two number",
    name="muliply"
)
```

```python
result = multiply_tool.invoke({"a":3,"b":5})
```

```python
print(result)
print(multiply_tool.name)
print(multiply_tool.args_schema)
print(multiply_tool.description)
```

# ***Using BaseTool Class***

```python
from langchain_core.tools import BaseTool
from typing import Type
```

```python
class MultiplyInput(BaseModel):
    a: int = Field(required=True, description="The first number to add")
    b: int = Field(required=True, description="The second number to add")
```

```python
class multiple_tools(BaseModel):
    name: str = "multiply"
    description: str = "Multiply two numbers"

    args_schema: Type[BaseModel] = MultiplyInput

    def _run(self, a: int, b: int) -> int:
        return a * b
```

```python
multiply_tool = multiple_tools()
```

```python
result = multiply_tool.invoke({'a':3, 'b':3})

print(result)
print(multiply_tool.name)
print(multiply_tool.description)

print(multiply_tool.args)
```

```python
from langchain_core.tools import tool

# Custom tools
@tool
def add(a: int, b: int) -> int:
    """Add two numbers"""
    return a + b

@tool
def multiply(a: int, b: int) -> int:
    """Multiply two numbers"""
    return a * b
```

```python
class MathToolkit:
    def get_tools(self):
        return [add, multiply]
```

```python
toolkit = MathToolkit()
tools = toolkit.get_tools()

for tool in tools:
    print(tool.name, "=>", tool.description)
```
